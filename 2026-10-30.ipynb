## Look for an exoplanet transit

- The NASA TESS space telescope needs telescopes on the ground to measure whether and when an exoplanet transit occurs.
- You look at data from one night in one filter and decide whether a transit occurred or not.
- You compare your conclusion with another person who has data from the same night, same star but different filter.
- THe two of you decide whether a transit happened.

## Measure the time a maximum of a pulsating variable star

- We pick a star other scientists have studied and measure something about it using data from one filter/one night.
- We likely measure what time the star reached its maximum brightness.

## Measure the timing of an eclipse in an eclipsing binary star or of a known exoplanet.

- Again, one filter/one night, measuring when some of the star's light is blocked.
