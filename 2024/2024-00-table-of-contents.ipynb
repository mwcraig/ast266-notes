# AST 266, Fall 2024 — Table of Contents

These are the 29 class-session notebooks from AST 266 in Fall 2024, from
August 26 through November 22, 2024.

The semester runs through two projects. The first half builds a color image of a Messier
object: pick a target, find it, plan when to observe it, calibrate the raw frames, align
them and combine them. The second half is about measuring light rather than making pictures
— aperture photometry, HR diagrams and distances, and finally the light curve of a
transiting exoplanet.

Each entry links to the notebook for that class day. Notebooks are grouped by topic below
and run in date order within each group.

*Note: a few notebooks carry a stale date in their own top heading (they were copied from a
previous year). The dates in this table come from the file names and are the correct ones.*
