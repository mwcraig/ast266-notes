### Choose an aperture radius

To choose an aperture radius, 

+ Choose a star for the seeing profile. It should NOT be too bright, or too close to another star, or too faint.
+ Save the aperture settings.
+ Make sure you see red and blue circles on the "Comparison Stars" tab
+ Save the variable information
+ Open one of the other images and record the FWHM of a star in that image
+ Repeat for the last of the three images.

Finally, choose a singe radius -- something like 1.5×FWHM is usuallly good.

Save one of the profile plots that supports your choce.
