# 2026-10-05 Patterns on the Main Sequence 

## And maybe stellar evolution

## Last time

+ Luminosity -- the amount of energy a star releases per second -- depends on the temperature of the star and its radius.
+ Flux -- the number of photons detected per second per unit area of the CCD sensor -- is what we can measure.

## Today

+ 90% of stars in the sky are on the *main sequence*.
+ All main sequence stars are fusing hydrogen into helium in their cores.
