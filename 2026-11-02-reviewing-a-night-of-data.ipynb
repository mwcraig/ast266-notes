## Review a night of data

This will give some experience with what a typical calibration image looks like.

1. Go to https://github.com/signup and create a free GitHub acount or log in if you already have an account
1. Go to the Launcher, then "Image galleries"
2. Open the images for your night in one tab. Open your github issue in another tab.
3. Start reviewing

## Night assignments are on the screen

## Steps

+ Write down the issue number for your night. You will need that later.
+ Copy/paste the link to the images of your night of data to a new browser window.
+ Click "here" in the line "Click here to edit README for this night"
+ Edit the README.
+ Save changes (one green button click)
+ Propose changes (another green button click)
+ Actually open the pull request (yet another green button click)
