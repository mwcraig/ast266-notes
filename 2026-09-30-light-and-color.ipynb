### What is the relationship between temperature and the spectrum of a star?

The higher temperature, the more blue light comapred to red light.

The lower the temperature, the longer the peak wavelength.

The higher the temperature, the higher the curve goes.

### What is the relationship  between temperature and B - V?

The higher the temperature, the lower the B-V color index.
