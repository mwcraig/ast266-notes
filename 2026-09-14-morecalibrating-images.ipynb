## Calibration recipe reminder

1. Subtract bias
    1. Combine several bias images to measure the bias
3. Subtract dark, scaled to match exposure time
    1. Combine several dark images to measure the dark current
5. Divide by flat
    1. Combine several flat images to measure the shadows.
    2. Scale the images so their typical value is 1.
